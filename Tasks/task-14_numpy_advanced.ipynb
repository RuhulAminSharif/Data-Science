###`Q-4:` Your are given an array which is havig some nan value. You job is to fill those nan values with most common element in the array.
```
arr=np.array([[1,2,np.nan],[4,2,6],[np.nan,np.nan,5]])

```

````python
arr=np.array([[1,2,np.nan],[4,2,6],[np.nan,np.nan,5]])

flatten = arr[ ~np.isnan(arr)].flatten().astype(np.int64)

most = np.bincount(flatten).argmax()

np.where( np.isnan(arr), most, arr)
````

### `Q-5:` Write a NumPy program

- to find the missing data in a given array. Return a boolean matrix.
- also try to fill those missing values with 0. For that, you can use `np.nan_to_num(a)`

```python
import numpy as np

np.array([[3, 2, np.nan, 1],
          [10, 12, 10, 9],
          [5, np.nan, 1, np.nan]])
```

````python
# code here
arr = np.array([[3, 2, np.nan, 1],
        [10, 12, 10, 9],
        [5, np.nan, 1, np.nan]])
bool_mat = np.isnan(arr)
print(bool_mat)
# one way
print(np.where(bool_mat, 0, arr))
#another way
print(np.nan_to_num(arr))
````

###`Q-7:` Plot this below equation.

$$\large y = \frac{e^x - e^{-x}}{e^x + e^{-x}}$$

**Note: This equation is called tanh activation function. In deep learning, many times this function is used. If you find some difference between the sigmoid function and this tanh function, note that to your notebook.**

````python
# code here
x = np.linspace(-10, 10, 42)
pos_exp = np.exp(x)
neg_exp = np.exp(-x)
tanh = (pos_exp - neg_exp) / (pos_exp+neg_exp)

import matplotlib.pyplot as plt
plt.plot(x, tanh)
````

### `Q-9:` Write a program implement Boradcasting Rule to check if two array can be added or not.
Given tuples of shapes.
```
shape of a- (3,2,2)
shape of b- (2,2)

check_broadcast(a, b) -> return Boolean (True if can broadcasted, False other wise.)
```

````python
# code here
def check_broadcast(shape_a, shape_b):
    list_a = list(shape_a)
    list_b = list(shape_b)
    list_a = list_a[::-1]
    list_b = list_b[::-1]
    for dim_a, dim_b in zip(list_a, list_b):
        if dim_a == dim_b or dim_a == 1 or dim_b == 1:
            continue
        else:
            return False
    return True

a = (3,2,2)
b = (2,2)
print(f"Can shapes {a} and {b} broadcast? {check_broadcast(a, b)}")

a = (3,1,2)
b = (2,2)
print(f"Can shapes {a} and {b} broadcast? {check_broadcast(a, b)}")

a = (3,2,2)
b = (3,1,2)
print(f"Can shapes {a} and {b} broadcast? {check_broadcast(a, b)}")

a = (3,2,5)
b = (2,2)
print(f"Can shapes {a} and {b} broadcast? {check_broadcast(a, b)}")
````
